# Data transformation using apply() & map() functions

```python
df= pd.read_excel("Movies_dataset.xlsx")
df
```

```python
# apply() function let you run the custom function on our dataset
df1 = df.copy()
df1['Revenue'] = df1['Revenue'].apply(lambda x : x * 2)  # it doubles the revenue
df1
```

```python
# Insert "Senior" if Salary > 100000 otherwise "Junior"

df1['Category'] = df1.apply(lambda row : "Senior" if  row['Revenue'] > 9.0 else "Junior", axis =1)
df1
```

```python
# map()
## map() function applies a custom function on the dataset
# It returns a scaler to every element of the  dataframe

# converted all the names in upper case letter
df1['Genre'] = df1['Genre'].map(lambda x : str(x).upper())
df1
```
